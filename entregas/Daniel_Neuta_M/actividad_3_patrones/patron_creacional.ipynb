### Justificación de la elección
Elegí el patrón **Factory Method** porque no quise hacer uso de un montón de condicionales para crear los objetos, sino armar un punto de extensión real usando la estructura de la GoF. Para aplicarlo, decidí diseñar un sistema de alertas para la universidad. Mi idea es que cuando pase algo en el campus, como una emergencia o la cancelación de una clase, yo pueda mandar notificaciones por correo o por mensaje de texto si es algo muy urgente.

Si me hubiera puesto a meter la instanciación de las clases de correo o SMS directamente en mi código principal, el programa me habría quedado súper amarrado y acoplado. El día de mañana que la universidad me pida añadir alertas por WhatsApp o Telegram, me tocaría entrar a romper y modificar el código. Con Factory Method solucioné esto: creé una clase base `NotificacionCreator` con un método de fábrica abstracto. Luego, se crean dos subclases (`EmailAlertCreator` y `SMSAlertCreator`) que heredan de esa base y son las que de verdad se encargan de fabricar el objeto específico. De esta manera, mi código principal solo da la orden de "mandar alerta" y se desentiende por completo de cómo se crea el objeto por debajo. Logré que mi código quedara súper flexible y estructurado como lo pide la guía.



```python
class Notificacion:
    """Interfaz Producto"""
    def enviar(self, mensaje: str) -> None:
        raise NotImplementedError("Subclases deben implementar este método")

class NotificacionEmail(Notificacion):
    """Producto Concreto A"""
    def enviar(self, mensaje: str) -> None:
        print(f"[EMAIL] Enviando correo institucional: {mensaje}")

class NotificacionSMS(Notificacion):
    """Producto Concreto B"""
    def enviar(self, mensaje: str) -> None:
        print(f"[SMS] Enviando mensaje de texto de urgencia: {mensaje}")

class NotificacionCreator:
    """Clase Creadora Base (Abstracta)"""
    def factory_method(self) -> Notificacion:
        """Este es el punto de extensión clave del patrón GoF"""
        raise NotImplementedError("Cada subclase debe definir su propio factory method")

    def enviar_alerta(self, mensaje: str) -> None:

        notificacion = self.factory_method()
        notificacion.enviar(mensaje)

class EmailAlertCreator(NotificacionCreator):
    """Creador Concreto A"""
    def factory_method(self) -> Notificacion:
        return NotificacionEmail()

class SMSAlertCreator(NotificacionCreator):
    """Creador Concreto B"""
    def factory_method(self) -> Notificacion:
        return NotificacionSMS()

```

```python
if __name__ == "__main__":
    print("--- Factory Method  ---\n")

    evento_grave = True

    if evento_grave:
        print("[Sistema] Detectado evento crítico en el campus.")
        emisor = SMSAlertCreator()
    else:
        print("[Sistema] Detectado aviso rutinario.")
        emisor = EmailAlertCreator()

    emisor.enviar_alerta("Clases suspendidas por mantenimiento eléctrico.")
```
