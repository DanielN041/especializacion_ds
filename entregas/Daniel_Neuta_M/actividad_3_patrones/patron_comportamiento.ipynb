## justificacion
Elegi el patron **strategy** porque Básicamente, el problema es que los métodos de pago cambian un montón; por ej, pagar todo con tarjeta de crédito te cobra un recargo, mientras que por PayPal la lógica es otra. Si metía todas esas reglas de negocio y cálculos dentro de la clase del carrito (CarritoMatricula), iba a terminar con un montón de condicionales que después seria tedioso mantener o corregir.Con Strategy logré separar cada lógica de cobro en su propia clase. Así, el carrito queda mas limpio, el código es más ordenado y, si el día de mañana la universidad quiere agregar otro método de pago, se puede hacer sin romper lo que ya funciona. Además, nos da la ventaja de cambiar la forma de cobro de forma dinámica en tiempo de ejecución según lo que elija el usuario.



```python
class CarritoMalo:
    def __init__(self):
        self.total = 0

    def agregar_materia(self, costo: int):
        self.total += costo

    def procesar_pago(self, metodo: str, datos: str):
        if metodo == "tarjeta":
            recargo = self.total * 0.05
            monto_final = self.total + recargo
            print(f" Cobrando ${monto_final} a la Tarjeta {datos} (Incluye 5% recargo).")
        elif metodo == "paypal":
            print(f"Redirigiendo a PayPal para pagar ${self.total} a: {datos}")
        else:
            print(" Metodo de pago no valido.")
```

```python
class EstrategiaPago:
    def pagar(self, monto: int) -> None:
        raise NotImplementedError()


class PagoTarjeta(EstrategiaPago):
    def __init__(self, numero_tarjeta: str):
        self.numero_tarjeta = numero_tarjeta

    def pagar(self, monto: int) -> None:
        recargo = monto * 0.05
        print(f"Cobro de ${monto + recargo} realizado con Tarjeta {self.numero_tarjeta} (Recargo aplicado).")

class PagoPayPal(EstrategiaPago):
    def __init__(self, email: str):
        self.email = email

    def pagar(self, monto: int) -> None:
        print(f"Login aprobado en PayPal. Pagado ${monto} desde la cuenta: {self.email}")

class CarritoMatricula:
    def __init__(self):
        self.total = 0
        self.estrategia: EstrategiaPago = None

    def agregar_materia(self, costo: int):
        self.total += costo

    def set_estrategia_pago(self, estrategia: EstrategiaPago):
        self.estrategia = estrategia

    def liquidar_matricula(self):
        if not self.estrategia:
            print("[Error] Selecciona un metodo de pago primero.")
            return
        self.estrategia.pagar(self.total)
```

```python
if __name__ == "__main__":
    # Prueba de la Version Sin Patron
    print("--- 1.sin implementación del patrón ---")
    carrito_malo = CarritoMalo()
    carrito_malo.agregar_materia(1000)
    carrito_malo.procesar_pago("tarjeta", "4557-XXXX-XXXX")
    print("-" * 50)


    print("\n--- 2. patron strategy---")
    carrito_bueno = CarritoMatricula()
    carrito_bueno.agregar_materia(500)
    carrito_bueno.agregar_materia(500) # Total: 1000

    print("-> Pago con Tarjeta")
    carrito_bueno.set_estrategia_pago(PagoTarjeta("4557-1234-9876"))
    carrito_bueno.liquidar_matricula()

    print("\n-> Cambio dinámico a PayPal")
    carrito_bueno.set_estrategia_pago(PagoPayPal("daniel.neuta@universidad.edu"))
    carrito_bueno.liquidar_matricula()
```
