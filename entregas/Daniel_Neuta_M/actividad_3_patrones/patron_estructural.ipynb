### Justificación de la elección
Elegi el patrón **Adapter** porque me parecio la forma más facil de entender como conectar dos clases que no encajan, usando un ejemplo cotidiano. 

 compre un celular nuevo que viene sin el entrada normal para los audífonos; solo tiene puerto **USB Tipo C**. Pero en mi casa tengo los audífonos de toda la vida, pero tienen el cable con el conector **Jack**. En el código, mi celular es el cliente y lo único que sabe hacer es buscar un puerto con el método `conectar_tipo_c()`, mientras que mis audífonos viejos solo entienden el método `conectar_pin_jack()`. Como lógicamente no puedo romper el celular para soldarle el cable viejo, ni voy a dañar los audífonos, la solución es meter una clase intermedia que sirva de puente: el **Adaptador** . Esta clase hereda de la interfaz para que el celular no ponga problema, pero por dentro agarra la señal y llama al método analógico de los audífonos viejos.




```python
class ConexiónTipoC:
    """La interfaz estándar que exige nuestro teléfono moderno"""
    def conectar_tipo_c(self) -> None:
        raise NotImplementedError("Debe implementar la conexión Tipo C")

class AudifonosViejosJack:
    """Clase externa o antigua. No podemos alterar su fabricación."""
    def conectar_pin_jack(self) -> None:
        print("[Audífonos Jack] Cable de 3.5mm conectado. Transmitiendo audio análogo...")


class AdaptadorJackATipoC(ConexiónTipoC):
    """El dongle/adaptador que compramos en la tienda.
    Se disfraza de Tipo C por fuera, pero maneja un Jack por dentro."""

    def __init__(self, audifonos_viejos: AudifonosViejosJack):
        self.audifonos_viejos = audifonos_viejos

    def conectar_tipo_c(self) -> None:
        print("[Adaptador] Convirtiendo señal digital Tipo C a análoga Jack...")
        # Traduce la llamada moderna al método que el componente viejo entiende
        self.audifonos_viejos.conectar_pin_jack()

class CelularModerno:

    def __init__(self, nombre_modelo: str):
        self.modelo = nombre_modelo

    def reproducir_musica(self, conexion: ConexiónTipoC):
        print(f"[{self.modelo}] Buscando dispositivo de audio Tipo C...")
        # El celular solo sabe usar la interfaz moderna
        conexion.conectar_tipo_c()
        print(f"[{self.modelo}] Reproduciendo mi playlist favorita con éxito.\n")

```

```python
if __name__ == "__main__":
    print("--- Patrón Adapter ---\n")


    mi_celular = CelularModerno("Galaxy S26 Ultra")


    audifonos_viejos = AudifonosViejosJack()


    dongle_adaptador = AdaptadorJackATipoC(audifonos_viejos)


    mi_celular.reproducir_musica(dongle_adaptador)
```
